## 0. Librerías


````python
import keras
from keras import layers
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
keras.utils.set_random_seed(SEED)

print(f"Keras version : {keras.__version__}")
print(f"Backend       : {keras.backend.backend()}")
````

# 1. Dataset: CIFAR-10

CIFAR-10 contiene 50,000 imágenes de entrenamiento y 10,000 de prueba. Las imágenes originales son de **32×32×3** y pertenecen a 10 categorías:

`airplane, automobile, bird, cat, deer, dog, frog, horse, ship, truck`.


````python
# Esta parte está resuelta.
(X_train, y_train), (X_test, y_test) = keras.datasets.cifar10.load_data()

y_train = y_train.squeeze()
y_test = y_test.squeeze()

class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

print("X_train original:", X_train.shape)
print("X_test original :", X_test.shape)
print("y_train         :", y_train.shape)
````

## 1.1 RGB → escala de grises

Usaremos la transformación:

$$Gray = 0.299R + 0.587G + 0.114B$$

Esta parte está resuelta porque el objetivo del ejercicio es practicar Keras.


````python
X_train = (
    0.299 * X_train[..., 0] +
    0.587 * X_train[..., 1] +
    0.114 * X_train[..., 2]
)

X_test = (
    0.299 * X_test[..., 0] +
    0.587 * X_test[..., 1] +
    0.114 * X_test[..., 2]
)

print("X_train grayscale:", X_train.shape)
print("X_test grayscale :", X_test.shape)
````

### Pregunta 1

Después de convertir las imágenes a escala de grises, cada imagen tiene tamaño `32×32`.

¿Cuántos valores tendrá cada imagen después de aplicar `Flatten()`?

**Respuesta:**

Cada imagen tendrá **32 × 32 = 1 024 valores**: `Flatten()` pone los píxeles de la matriz uno detrás de otro en un solo vector de 1 024 posiciones.

## 1.2 Normalización

Los valores de los píxeles están entre 0 y 255. Completa el código para llevarlos al rango `[0,1]`.


````python
# TODO 1 — Normalización
X_train = X_train.astype("float32") / 255.0
X_test  = X_test.astype("float32") / 255.0

print(X_train.min(), X_train.max())
````

````python
# Validación TODO 1
assert X_train.dtype == np.float32
assert X_test.dtype == np.float32
assert X_train.min() >= 0.0
assert X_train.max() <= 1.0
print("✓ Datos normalizados correctamente")
````

## 1.3 Visualización

Esta parte está resuelta.


````python
plt.figure(figsize=(12, 6))
for i in range(12):
    plt.subplot(3, 4, i + 1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(class_names[y_train[i]])
    plt.xticks([])
    plt.yticks([])
plt.tight_layout()
plt.show()
````

---
# 2. Construir la red neuronal

Construiremos la misma arquitectura conceptual vista en clase:

```text
Imagen 32×32
   ↓
Flatten
   ↓
1024 valores
   ↓
Dense(128)
   ↓
ReLU
   ↓
Dense(10)
   ↓
Logits
```

> La última capa **no** debe tener Softmax. Trabajaremos con logits.


````python
# TODO 2 — Completa la arquitectura
model = keras.Sequential([
    keras.Input(shape=(32, 32)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10)
])

model.summary()
````

````python
# Validación TODO 2
expected_params = (1024 * 128 + 128) + (128 * 10 + 10)
assert model.count_params() == expected_params
print(f"✓ Arquitectura correcta: {model.count_params():,} parámetros")
````

### Preguntas

**2.** ¿Por qué necesitamos `Flatten()` antes de la primera capa `Dense`?

**Respuesta:**

Porque una capa `Dense` espera como entrada un **vector** (una dimensión por ejemplo), y una imagen en escala de grises es una **matriz** de 32×32. `Flatten()` convierte esa matriz en un vector de 1 024 valores para poder conectarla con todas las neuronas de la capa densa.

**3.** ¿Por qué la última capa tiene 10 neuronas?

**Respuesta:**

Porque hay **10 clases** (airplane, automobile, bird, cat, deer, dog, frog, horse, ship, truck). Cada neurona de salida produce un logit asociado a una clase, y la clase predicha es la de mayor valor.

**4.** ¿Qué función cumple `ReLU`?

**Respuesta:**

`ReLU(x) = max(0, x)` introduce una **no linealidad**. Sin ella, apilar capas `Dense` equivaldría a una sola transformación lineal y la red solo podría separar clases con fronteras lineales. Con ReLU puede aprender relaciones más complejas; además es muy barata de calcular y ayuda a que los gradientes no se desvanezcan.

---
# 3. Configurar el entrenamiento

Usaremos:

- `Adam`
- `SparseCategoricalCrossentropy`
- `accuracy`

Como la red produce logits, la pérdida debe configurarse con `from_logits=True`.


````python
lr_rate = 0.001

# TODO 3 — Configura el entrenamiento
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=lr_rate),
    loss=keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=["accuracy"]
)
````

### Pregunta 5

¿Por qué usamos `from_logits=True`?

**Respuesta:**

Porque la última capa **no tiene Softmax**: entrega logits (valores reales sin normalizar). Con `from_logits=True` la función de pérdida aplica el softmax internamente, de forma numéricamente más estable (usa la formulación log-sum-exp) que calcular primero el softmax y después el logaritmo.

---
# 4. Entrenar el modelo

Usaremos 10 épocas y mini-batches de 64 ejemplos. Completa `fit()`.


````python
epochs = 10
batch_size = 64

# TODO 4 — Entrenamiento
# (Este notebook usa el conjunto de prueba como validation_data, porque no hay un split de validación aparte.)
history = model.fit(
    X_train,
    y_train,
    epochs=epochs,
    batch_size=batch_size,
    validation_data=(X_test, y_test),
    shuffle=True
)
````

### Preguntas

**6.** ¿Qué representa una `epoch`?

**Respuesta:**

Una **epoch** es una pasada completa por todo el conjunto de entrenamiento: el modelo ve cada una de las 50 000 imágenes exactamente una vez.

**7.** ¿Qué representa `batch_size=64`?

**Respuesta:**

Que los pesos se actualizan después de procesar **64 ejemplos** (un mini-batch), no con cada imagen ni con todo el dataset. Con 50 000 imágenes hay ≈ 782 actualizaciones de pesos por epoch.

## 4.1 Curvas de aprendizaje

Esta parte está resuelta.


````python
plt.figure(figsize=(7, 4))
plt.plot(history.history["accuracy"], label="Train accuracy")
plt.plot(history.history["val_accuracy"], label="Validation accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Evolución del entrenamiento")
plt.legend()
plt.show()
````

### Pregunta 8

Observa las dos curvas. ¿Hay evidencia de overfitting? Justifica brevemente.

**Respuesta:**

No hay evidencia clara de overfitting. En la época 10, el accuracy de entrenamiento es ≈ 38.3 % y el de validación ≈ 36.4 %: la brecha es de solo ≈ 1.8 puntos y ambas curvas suben juntas durante las 10 épocas. La pérdida de validación también sigue bajando (de 1.98 en la época 1 a 1.78 en la 10) y nunca sube mientras la de entrenamiento baja, que sería la señal típica de overfitting.

Lo que se observa más bien es que la red **aprende despacio y con poca capacidad** (el accuracy de entrenamiento es bajo, ≈ 38 %), es decir, tiende a quedarse corta (underfitting) más que a memorizar.

# 5. Evaluar el modelo

Completa `evaluate()`.


````python
# TODO 5 — Evaluación
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    batch_size=batch_size,
    verbose=0
)

print(f"Test loss     : {test_loss:.4f}")
print(f"Test accuracy : {test_accuracy:.4f}")
print(f"Accuracy (%)  : {test_accuracy * 100:.2f}%")
````

# 6. Logits y Softmax

La salida de la red son 10 logits. Para interpretarlos como probabilidades necesitamos Softmax.


````python
index = 0
image = X_test[index]
true_class = y_test[index]

plt.imshow(image, cmap="gray")
plt.title(f"Clase real: {class_names[true_class]}")
plt.axis("off")
plt.show()

image_batch = np.expand_dims(image, axis=0)
logits = model(image_batch, training=False)

print("Logits:")
print(keras.ops.convert_to_numpy(logits))
````

````python
# TODO 6 — Logits -> probabilidades
probabilities = keras.ops.softmax(logits)

# TODO 7 — Clase con mayor probabilidad
prediction = keras.ops.argmax(probabilities, axis=1)
prediction = int(keras.ops.convert_to_numpy(prediction)[0])

print("Clase real     :", class_names[true_class])
print("Clase predicha :", class_names[prediction])
````

````python
probabilities_np = keras.ops.convert_to_numpy(probabilities)[0]

print("\nProbabilidades:")
for i, p in enumerate(probabilities_np):
    marker = " ← predicción" if i == prediction else ""
    print(f"{class_names[i]:12s}: {p:.4f}{marker}")

print("\nSuma:", probabilities_np.sum())
````

### Pregunta 10

¿Por qué las probabilidades producidas por `Softmax` deben sumar aproximadamente 1?

**Respuesta:**

Porque Softmax convierte los 10 logits en una **distribución de probabilidad** sobre clases mutuamente excluyentes: cada salida es `exp(z_i) / Σ exp(z_j)`, es positiva y, por construcción, la suma de todas es 1 (la imagen pertenece a una y solo una de las clases). La suma sale «aproximadamente» 1 por el redondeo en coma flotante.

### 3. Después de `Flatten()`, la imagen se convierte en un vector de 1024 valores. ¿Qué información importante sobre la imagen podría perderse al tratarla únicamente como un vector?

**Respuesta:**

Se pierde la **estructura espacial**: la red no sabe qué píxeles eran vecinos (arriba/abajo/izquierda/derecha), así que no "ve" bordes, texturas ni formas locales; cada píxel es solo una entrada más. Si se reordenaran los píxeles de todas las imágenes de la misma manera, una red `Dense` aprendería igual de bien (o mal), lo que muestra que no explota la vecindad. Tampoco es invariante a traslaciones: un objeto desplazado unos píxeles produce un vector muy distinto.

### 4. ¿Qué tipo de arquitectura podría aprovechar mejor la estructura espacial de las imágenes?

**Respuesta:**

Las **redes neuronales convolucionales (CNN)**: usan filtros pequeños que se desplazan por la imagen y comparten pesos, de modo que detectan patrones locales (bordes, texturas, partes de objetos), conservan la relación entre píxeles vecinos, necesitan muchos menos parámetros y son más robustas a que el objeto cambie de posición.
